```python
import sqlite3
import pandas as pd
import random

random.seed(42)

conn = sqlite3.connect("universidad.db")

materias = [
    "Python",
    "Estadística",
    "Base de Datos",
    "Matemáticas",
    "Contabilidad",
    "Administración",
    "Investigación",
    "Inglés",
    "Programación",
    "Auditoría"
]

datos_notas = []

for i in range(1, 5001):
    id_estudiante = random.randint(1, 5000)
    materia = random.choice(materias)
    nota = round(random.uniform(5, 10), 2)
    asistencia = random.randint(60, 100)

    datos_notas.append([
        i,
        id_estudiante,
        materia,
        nota,
        asistencia
    ])

df_notas = pd.DataFrame(datos_notas, columns=[
    "id_nota",
    "id_estudiante",
    "materia",
    "nota",
    "asistencia"
])

df_notas.to_sql("notas", conn, if_exists="replace", index=False)

conn.close()

print("Base de datos universidad.db creada con 5000 registros.")
```
