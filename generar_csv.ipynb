```python
import pandas as pd
import random

random.seed(42)

nombres = ["Ana", "Luis", "Carlos", "María", "José", "Camila", "Pedro", "Daniela", "Andrés", "Sofía"]
apellidos = ["Pérez", "Gómez", "Torres", "López", "Castillo", "Díaz", "Morales", "Ramírez", "Vega", "Ruiz"]
carreras = ["Sistemas", "Administración", "Contabilidad", "Derecho", "Psicología", "Educación"]
generos = ["Masculino", "Femenino"]

datos = []

for i in range(1, 5001):
    nombre = random.choice(nombres) + " " + random.choice(apellidos)
    carrera = random.choice(carreras)
    edad = random.randint(18, 35)
    genero = random.choice(generos)
    semestre = random.randint(1, 10)

    datos.append([i, nombre, carrera, edad, genero, semestre])

df = pd.DataFrame(datos, columns=[
    "id_estudiante",
    "nombre",
    "carrera",
    "edad",
    "genero",
    "semestre"
])

df.to_csv("estudiantes.csv", index=False, encoding="utf-8")

print("CSV estudiantes.csv creado con 5000 registros.")
```
